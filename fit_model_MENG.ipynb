```python
def trace_generique(Y, Ymod, modele_name):
    n = int(max(Ymod.max(), Y.max())) + 5

    # Graph sizee
    plt.figure(figsize=(8,8))
    plt.xlim(0, n)
    plt.ylim(0, n)

    # Scatterplot and legend
    scatter = plt.scatter(Ymod, Y, c=labels_nbdd, marker = '.')
    legend = plt.legend(*scatter.legend_elements(), title='n° BDD')
    plt.gca().add_artist(legend)

    # line y=x et line + ou - 20% ou 40%
    x_line, y_line = np.arange(n), np.arange(n)
    y_line_20plus = y_line * 1.2
    y_line_20minus = y_line * 0.8
    y_line_40plus = y_line * 1.4
    y_line_40minus = y_line * 0.6
    plt.plot(x_line, y_line, color = 'black')
    plt.plot(x_line, y_line_20plus, linestyle = '--', color = 'black')
    plt.plot(x_line, y_line_20minus, linestyle = '--', color = 'black')
    plt.plot(x_line, y_line_40plus, linestyle = ':', color = 'black')
    plt.plot(x_line, y_line_40minus, linestyle = ':', color = 'black')

    plt.xlabel("f_c predic (MPa)")
    plt.ylabel("f_c exp (MPa)")
    plt.title("Model : " + modele_name)
    plt.grid()
    plt.show()

def trace_modele_ML(modele, X, Y, modele_name = None):
    Ymod = modele.predict(X)
    if modele_name is None :
        modele_name = str(type(modele))
    trace_generique(Y, Ymod, modele_name)
```
